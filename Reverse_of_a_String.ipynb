```python
#برنامه ای می نویسیم که یک رشته را دریافت بررسی میکنید که ایا این رشته با برعکس خودش یکسان است و یا خیر
def reversed_str():
  str = input()
  l = len(str)
  if l:
    i = 0
    j = l -1
    while i < j:
      if str[i] != str[j]:
        return False
      i += 1
      j -= 1
    return True
  return True
```
