Find the prime number just before and after the number which is given

```python
def is_prime(n):
  if n <= 1:
    return False
  for i in range(2, int(n**0.5) + 1):
    if n % i == 0:
      return False
  return True
```

```python
def find_prime(x, d):
  x += d
  while not is_prime(x):
    if x <= 1:
      return "No prime number found"
    x += d
  return x
```

```python
def prime_numbers(x):
  x = round(abs(x))
  if is_prime(x):
    print(x)
    print(find_prime(x, -1))
  else:
    print(find_prime(x, -1))
    print(find_prime(x, 1))
```

```python
prime_numbers(5)
```

```python
prime_numbers(9)
```

```python
prime_numbers(15.7)
```

```python
prime_numbers(-15.7)
```
