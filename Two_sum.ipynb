```python
# برنامه ای مینویسیم که یک لیست و یک عدد دریافت شماره ایندکس دو خانه متقاوت در لیست که مجموع مفدار داخلی انها برابر با عدد ورودی است را بر میگرداند.
def two_sum(nums, target):
  _dict = {}
  for key, value in enumerate(nums):
    _val = target - value
    if _val in _dict:
      return [key, _dict[_val]]
    else:
      _dict[value] = key
```

```python
_nums = [4, 1, 3, 2, 5]
target = 8
two_sum(_nums, target)
```
