```python
#یک لیست مرتب را دریافت تمامی اعداد داخل ان را به توان دو مرساند و در نهایت یک لیست مرتب را بر میگرداند
def sorted_squares(_list):
  _lst = []
  i = 0
  j = len(_list) - 1
  while i <= j:
    if ((_list[i])**2) > ((_list[j])**2):
      _lst.append((_list[i])**2)
      i += 1
    else:
      _lst.append((_list[j])**2)
      j -= 1
  return _lst
```

```python
sorted_squares([-4, -2, 1, 2, 3, 4])
```
