```python
#برنامه ای مینویسیم که تعداد حروف کوچک داخل یک رشته را برمیگرداند
def  number_of_small_letters():
  _str = input()
  counter = 0
  for i in _str:
    if ord(i) in range(97, 123):
      counter += 1
  return(counter)

```

```python
number_of_small_letters()
```
