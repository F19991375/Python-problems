```python
#Write a Python program that takes a string of 1s and 0s and returns the longest
#subsequence that has repeated at least once, all without using sorting algorithms.
_str = input()
i = j = 0
n = len(_str)
longest_substr = _str[0]
cur_str = [_str[0]]
while i < n:
  if "".join(cur_str) in _str[j+1:]:
    if len(longest_substr) < len(''.join(cur_str)):
      longest_substr = ''.join(cur_str)
    j += 1
    cur_str.append(_str[j])
  else:
    cur_str.pop(0)
    i += 1
print(longest_substr)
```
